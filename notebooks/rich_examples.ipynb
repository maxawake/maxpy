```python
import time

from maxpy.logger import add_file_handler, logging

add_file_handler(logging.getLogger(__name__))
import time

from rich.progress import Progress, SpinnerColumn

import maxpy.logger.notebook
```

```python

logger = logging.getLogger(__name__)
logger.critical("Processing %d snapshots", 128)

array = range(10)  # Example array for demonstration purposes

with Progress(SpinnerColumn(), *Progress.get_default_columns()) as progress:
    for item in progress.track(array, description="Processing"):
        time.sleep(0.3)
```

```python
from rich.console import Console
from rich.table import Table

console = Console()

console.print("[bold green]Analysis started[/]")
console.print({"snapshots": 128, "resolution": [256, 256, 256]})

table = Table("Run", "Status", title="Simulation jobs")
table.add_row("run_01", "[green]Finished[/]")
table.add_row("run_02", "[yellow]Running[/]")
console.print(table)
```

```python
import json

import yaml
from rich.console import Console
from rich.panel import Panel
from rich.pretty import Pretty

console = Console()
config = {"grid": [256, 256, 256], "solver": {"tolerance": 1e-8, "max_iterations": 1000}}

console.print("[bold green]Ready to start[/]")
console.print(Panel.fit(yaml.safe_dump(config, indent=4), title="Configuration", border_style="cyan"))
```

```python
from time import sleep

from rich.progress import Progress, SpinnerColumn, TextColumn

columns = [SpinnerColumn(), *Progress.get_default_columns(), TextColumn("residual={task.fields[residual]:.2e}")]

with Progress(*columns, console=console) as progress:
    task = progress.add_task("Solving", total=50, residual=1.0)
    for step in range(1, 51):
        sleep(0.1)
        progress.update(task, advance=1, residual=0.8 ** step, refresh=True)
```
